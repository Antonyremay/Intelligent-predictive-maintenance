```python
test_fd001=pd.read_csv(r'C:\Users\antony remane\OneDrive\Desktop\ML - Prac\Intelligent Predictive Maintenance System\CMAPSSData\test_FD001.txt',sep=r'\s+',header=None)
rul_fd001=pd.read_csv(r'C:\Users\antony remane\OneDrive\Desktop\ML - Prac\Intelligent Predictive Maintenance System\CMAPSSData\RUL_FD001.txt',sep=r'\s+',header=None)
test_fd001.columns=column_names
rul_fd001.columns=['RUL']
print(test_fd001.head())
print(test_fd001.shape)
test_fd001.info()
print(rul_fd001.head())
print(rul_fd001.shape)
rul_fd001.info()
```
